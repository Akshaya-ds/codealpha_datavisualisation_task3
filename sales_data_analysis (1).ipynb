```python
#now we need to do Data Cleaning
#we will check if there are any missing values
df.isnull().sum()
#Addressline,State,Territory are having null values
```

```python
df.fillna(0,inplace=True)
#we are  going to fill all the missing values with 0 so that there will be no changes in data
```

```python
df.isnull().sum()
#now there are no missing values
```

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
#this will stores the data with removed duplicates
```

```python
df["SALES"].sum()
#finding the total revenue
#the total revenue is 1,00,32,628
```

```python
#I want to find which product is sold most
df.groupby("PRODUCTLINE")["SALES"].sum().sort_values(ascending=False)
#Insight : Classic Cars generated highest revenue
```

```python
df.groupby("PRODUCTLINE")["SALES"].sum().sort_values(ascending=False).plot(kind="bar",color="red")
plt.title("REVENUE OF EACH PRODUCT")
plt.xlabel("PRODUCT NAMES")
plt.ylabel("SALES/REVENUE")
plt.show()
```

```python
#top 10 countries sales
df.groupby("COUNTRY")["SALES"].sum().sort_values(ascending=False).head(10)
#Insight : USA has made more sales
```

```python
df.groupby("COUNTRY")["SALES"].sum().sort_values(ascending=False).head(10).plot(kind="bar")
plt.title("TOP 10 COUNTRIES SALES")
plt.xlabel("COUNTRY NAMES")
plt.ylabel("COUNTRY SALES")
plt.show()
```

```python
#now we will find out in which month sales are highest
monthly_sales=df.groupby("MONTH_ID")["SALES"].sum().sort_values(ascending=False)
print(monthly_sales)
#Insight : November is having highest sales
```

```python
monthly_sales.plot(kind="line",marker="o")
plt.title("MONTHLY SALES")
plt.xlabel("MONTH")
plt.ylabel("SALES")
plt.xticks(range(1,13)) #used to change the scale of graph
plt.show()
```

```python
#finding out which year had highest revenue
yearly_sales=df.groupby("YEAR_ID")["SALES"].sum().sort_values(ascending=False)
print(yearly_sales)
#Insight : 2004 is having highest revenue
```

```python
yearly_sales.plot(kind="bar")
plt.title("YEARLY SALES")
plt.xlabel("YEAR")
plt.ylabel("SALES")
plt.show()
```

```python
df.groupby("DEALSIZE")["SALES"].sum()
```

```python
df.groupby("DEALSIZE")["SALES"].sum().plot(kind="pie", autopct="%1.1f%%")

plt.title("Sales by Deal Size")

plt.show()
```

```python
# now lets find top customers
df.groupby("CUSTOMERNAME")["SALES"].sum().sort_values(ascending=False).head(10)
#Insight : Euro Shopping channel is the top most customer with highest purchasing
```

```python
df.groupby("CUSTOMERNAME")["SALES"].sum().sort_values(ascending=False).head(10).plot(kind="bar")
plt.title("TOP 10 CUSTOMERS")
plt.xlabel("CUSTOMER NAMES")
plt.ylabel("SALES")
plt.show
```
